```python
# STEP FINAL 1 — VERIFIKASI HASIL EKSPERIMEN

import pandas as pd
import numpy as np

ml_cv = pd.read_csv("results/ml_cv.csv")
ann_cv = pd.read_csv("results/ann_cv.csv")
test_final = pd.read_csv("results/test_final.csv")

print("=== JUMLAH HASIL EKSPERIMEN ===")
print("ML CV       :", len(ml_cv))
print("ANN CV      :", len(ann_cv))
print("Final Test  :", len(test_final))

print("\n=== PEMERIKSAAN MISSING VALUE ===")

for name, df in [
    ("ML CV", ml_cv),
    ("ANN CV", ann_cv),
    ("Final Test", test_final)
]:
    print(f"{name}: {df.isna().sum().sum()} missing values")

print("\n=== PEMERIKSAAN METRIK FINAL ===")

checks = pd.DataFrame({
    "Pemeriksaan": [
        "Jumlah hasil final = 12",
        "Tidak ada metrik kosong",
        "MAE >= 0",
        "RMSE >= MAE",
        "R2 <= 1",
        "Tidak ada kombinasi split-target-family duplikat",
        "Semua kombinasi split-target-family tersedia"
    ],
    "Lulus": [
        len(test_final) == 12,
        test_final[["MAE", "RMSE", "R2"]].notna().all().all(),
        (test_final["MAE"] >= 0).all(),
        (test_final["RMSE"] >= test_final["MAE"]).all(),
        (test_final["R2"] <= 1).all(),
        not test_final.duplicated(
            ["split", "target", "family"]
        ).any(),
        (
            set(
                map(
                    tuple,
                    test_final[
                        ["split", "target", "family"]
                    ].to_numpy()
                )
            )
            ==
            {
                (s, t, f)
                for s in ["80_20", "70_30", "temporal"]
                for t in ["CO", "NOX"]
                for f in ["ML", "ANN"]
            }
        )
    ]
})

display(checks)

print("\n=== HASIL PENGUJIAN AKHIR ===")
display(
    test_final.sort_values(
        ["target", "split", "RMSE"]
    ).round(4)
)
```

```python
# STEP FINAL 3 — ANALISIS GENERALIZATION GAP

import pandas as pd
import numpy as np

# Membaca hasil eksperimen
ml_cv = pd.read_csv("results/ml_cv.csv")
ann_cv = pd.read_csv("results/ann_cv.csv")
test_final = pd.read_csv("results/test_final.csv")

# 1. Model ML terbaik berdasarkan RMSE CV
best_ml = (
    ml_cv
    .sort_values("RMSE_mean")
    .drop_duplicates(["split", "target"])
    .copy()
)

best_ml["family"] = "ML"

# 2. Model ANN terbaik berdasarkan RMSE CV
best_ann = (
    ann_cv
    .sort_values("RMSE_mean")
    .drop_duplicates(["split", "target"])
    .copy()
)

best_ann["family"] = "ANN"

# 3. Gabungkan hasil CV
cv_summary = pd.concat(
    [
        best_ml[["split", "target", "family", "RMSE_mean"]],
        best_ann[["split", "target", "family", "RMSE_mean"]]
    ],
    ignore_index=True
)

# 4. Gabungkan dengan hasil testing
generalization = test_final.merge(
    cv_summary,
    on=["split", "target", "family"],
    how="left",
    validate="one_to_one"
)

# 5. Hitung generalization gap
generalization["RMSE_Gap"] = (
    generalization["RMSE"] - generalization["RMSE_mean"]
)

generalization["RMSE_Ratio"] = (
    generalization["RMSE"] / generalization["RMSE_mean"]
)

# 6. Tampilkan hasil
print("=== ANALISIS GENERALIZATION GAP ===")

display(
    generalization[
        [
            "split",
            "target",
            "family",
            "RMSE_mean",
            "RMSE",
            "RMSE_Gap",
            "RMSE_Ratio",
            "R2"
        ]
    ]
    .sort_values(["target", "split", "family"])
    .round(4)
)
```

```python
#Contoh menjawab dan membuktikan sebuah pertanyaan
#Mengapa kamu memilih Random Forest Regression untuk memprediksi emisi NOx, bukan Linear Regression?
#Apa perbedaan cara kerja Linear Regression dan Random Forest Regression?
#Mengapa hubungan antara parameter operasi turbin gas dan emisi NOx mungkin lebih cocok dimodelkan menggunakan Random Forest?
#Apa kelemahan Random Forest yang perlu diperhatikan dalam penelitian ini?
#step 1 Membaca dan menggabungkan dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

data_dir = Path("data")

files = sorted(data_dir.glob("*.csv"))

print("Jumlah file:", len(files))
print("File ditemukan:", [f.name for f in files])

if len(files) != 5:
    raise ValueError("Diharapkan 5 file CSV dalam folder data.")

df = pd.concat(
    [pd.read_csv(file) for file in files],
    ignore_index=True
)

print("\nUkuran dataset:", df.shape)

display(df.head())
```

```python
#step 2 menentukan fitur dan target
features = [
    "AT", "AP", "AH", "AFDP",
    "GTEP", "TIT", "TAT", "TEY", "CDP"
]

target = "NOX"

X = df[features]
y = df[target]

print("Jumlah fitur:", X.shape[1])
print("Jumlah sampel:", X.shape[0])
print("Target:", target)

display(X.head())
```

```python
#step 3 membagi trening dan testing
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training:", X_train.shape)
print("Testing :", X_test.shape)
```

```python
# step 4 membuat kedua model
#menggunakan pipeline agar preprocessing dilakukan secara benar dan tidak menyebabkan data leakage
#
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

linear_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])

rf_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])

print("Linear Regression siap.")
print("Random Forest Regression siap.")
```

```python
#step 5 Melakukan 5-Fold Cross Validation
from sklearn.model_selection import KFold, cross_validate

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

models = {
    "Linear Regression": linear_model,
    "Random Forest": rf_model
}

cv_results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=1
    )

    cv_results.append({
        "Model": name,
        "MAE_mean": -scores["test_MAE"].mean(),
        "MAE_std": scores["test_MAE"].std(),
        "RMSE_mean": -scores["test_RMSE"].mean(),
        "RMSE_std": scores["test_RMSE"].std(),
        "R2_mean": scores["test_R2"].mean(),
        "R2_std": scores["test_R2"].std()
    })

df_cv = pd.DataFrame(cv_results)

display(df_cv.round(4))
```

```python
#step 6 training dan evaluasi akhir 
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

test_results = []
predictions = {}

for name, model in models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    predictions[name] = y_pred

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    test_results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

df_test = pd.DataFrame(test_results)

print("=== HASIL PENGUJIAN AKHIR NOX ===")

display(df_test.round(4))
```

```python
#step 7 visualisasi hasil aktual vs prediksi 
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, (name, y_pred) in zip(axes, predictions.items()):

    ax.scatter(
        y_test,
        y_pred,
        alpha=0.3,
        s=12
    )

    lower = min(y_test.min(), y_pred.min())
    upper = max(y_test.max(), y_pred.max())

    ax.plot(
        [lower, upper],
        [lower, upper],
        "r--",
        label="Prediksi ideal"
    )

    ax.set_title(name)
    ax.set_xlabel("NOx Aktual (ppm)")
    ax.set_ylabel("NOx Prediksi (ppm)")
    ax.legend()
    ax.grid(alpha=0.3)

plt.suptitle("Perbandingan Prediksi NOx")
plt.tight_layout()
plt.show()
```

```python
#step 8 visualisasi residual 
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, (name, y_pred) in zip(axes, predictions.items()):

    residual = y_test.to_numpy() - y_pred

    ax.scatter(
        y_pred,
        residual,
        alpha=0.3,
        s=12
    )

    ax.axhline(
        0,
        color="red",
        linestyle="--"
    )

    ax.set_title(name)
    ax.set_xlabel("NOx Prediksi (ppm)")
    ax.set_ylabel("Residual (ppm)")
    ax.grid(alpha=0.3)

plt.suptitle("Analisis Residual")
plt.tight_layout()
plt.show()
```

```python
#split 70:30, 5-fold cross-validation, dan mengevaluasi seluruh prediksi dalam satuan CO asli.
# step 1 Membaca dataset dan memisahkan training-testing
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import train_test_split

# Membaca dataset
files = sorted(Path("data").glob("*.csv"))

if len(files) != 5:
    raise ValueError(f"Ditemukan {len(files)} CSV, seharusnya 5.")

df = pd.concat(
    [pd.read_csv(f) for f in files],
    ignore_index=True
)

features = [
    "AT", "AP", "AH", "AFDP",
    "GTEP", "TIT", "TAT", "TEY", "CDP"
]

X = df[features]
y = df["CO"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42
)

print("Total observasi :", len(df))
print("Data training   :", len(X_train))
print("Data testing    :", len(X_test))
print("Skewness training CO :", round(y_train.skew(), 4))
```

```python
#step 2 Membuktikan skewness dan efek log
if (y_train < 0).any():
    raise ValueError("Transformasi log1p memerlukan CO >= 0.")

co_log = np.log1p(y_train)

print("Skewness CO asli :", round(y_train.skew(), 4))
print("Skewness log CO  :", round(co_log.skew(), 4))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(y_train, bins=60, edgecolor="black")
axes[0].set_title("CO Asli — Training")
axes[0].set_xlabel("CO")

axes[1].hist(co_log, bins=60, edgecolor="black")
axes[1].set_title("log1p(CO) — Training")
axes[1].set_xlabel("log1p(CO)")

plt.tight_layout()
plt.show()
```

```python
# step 3 Membangun dua model SVR
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.compose import TransformedTargetRegressor
from sklearn.base import clone

svr_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("svr", SVR(
        kernel="rbf",
        C=10,
        epsilon=0.1,
        gamma="scale"
    ))
])

model_asli = clone(svr_pipeline)

model_log = TransformedTargetRegressor(
    regressor=clone(svr_pipeline),
    func=np.log1p,
    inverse_func=np.expm1
)

models = {
    "SVR tanpa log": model_asli,
    "SVR dengan log": model_log
}

print("Kedua model berhasil disiapkan.")
```

```python
# step 4 Melakukan 5-fold cross-validation
from sklearn.model_selection import KFold, cross_validate

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

cv_results = []

for name, model in models.items():

    print("Evaluasi:", name)

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=1
    )

    cv_results.append({
        "Model": name,
        "MAE_mean": -scores["test_MAE"].mean(),
        "RMSE_mean": -scores["test_RMSE"].mean(),
        "RMSE_std": scores["test_RMSE"].std(),
        "R2_mean": scores["test_R2"].mean()
    })

df_cv = pd.DataFrame(cv_results)

display(df_cv.round(4))
```

```python
#step 5 Memilih model berdasarkan cross-validation
best_row = df_cv.loc[df_cv["RMSE_mean"].idxmin()]
best_name = best_row["Model"]

print("Model terbaik berdasarkan CV RMSE:")
print(best_name)

print("\nRMSE CV :", round(best_row["RMSE_mean"], 4))
print("R2 CV   :", round(best_row["R2_mean"], 4))
```

```python
#step 6 Evaluasi pada data testing
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

best_model = models[best_name]

best_model.fit(X_train, y_train)

y_pred = best_model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("=== HASIL FINAL PREDIKSI CO ===")
print("Model :", best_name)
print("MAE   :", round(mae, 4))
print("RMSE  :", round(rmse, 4))
print("R2    :", round(r2, 4))
```

```python
# step 7 visualisasi prediksi co
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, (name, model) in zip(axes, models.items()):

    # Melatih model menggunakan data training
    model.fit(X_train, y_train)

    # Prediksi CO pada skala asli
    y_pred_model = model.predict(X_test)

    # Menghitung evaluasi
    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred_model)
    )

    r2 = r2_score(y_test, y_pred_model)

    # Scatter plot
    ax.scatter(
        y_test,
        y_pred_model,
        alpha=0.3,
        s=12
    )

    # Garis prediksi ideal
    lower = min(y_test.min(), y_pred_model.min())
    upper = max(y_test.max(), y_pred_model.max())

    ax.plot(
        [lower, upper],
        [lower, upper],
        "r--",
        label="Prediksi ideal"
    )

    ax.set_title(
        f"{name}\nRMSE = {rmse:.4f} | R² = {r2:.4f}"
    )

    ax.set_xlabel("CO Aktual")
    ax.set_ylabel("CO Prediksi")
    ax.legend()
    ax.grid(alpha=0.3)

plt.suptitle(
    "Perbandingan Prediksi CO: SVR Tanpa Log vs Dengan Log",
    fontsize=13
)

plt.tight_layout()
plt.show()
```
